```python
## Step3: Select the first seven rows
```

```python
asia_cars = df[df["origin"] == "Asia"]
selected_columns = asia_cars[["vehicle_weight", "model_year"]]
first_seven = selected_columns.head(7)

print("Asian cars:", len(asia_cars))
display(first_seven)
```

```python
## convert to numpy array 
```

```python
X = first_seven.to_numpy()
X
print(X.shape)
```

```python
X.shape
```

```python
## Multiple the transpose by X
```

```python
XTX = X.T @ X
print(XTX)
```

```python
## Step 6: Invert XTX
```

```python
XTX_inverse = np.linalg.inv(XTX)

print(XTX_inverse)
```

```python
## Create an array y
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

y
```

```python
## Step 8: Calculate w
```

```python
w = XTX_inverse @ X.T @ y

w
```

```python
##  Step 9: Calculate the sum
```

```python
w.sum()
```

```python
round(w.sum(), 3)
```
